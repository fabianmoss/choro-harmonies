```python
# a simple script that removes existing chords from musicxml files

import music21 as m21
from pathlib import Path

output_dir = Path("choro_model_ft/scores")
output_dir.mkdir(parents=True, exist_ok=True)

for score_path in Path("choro_corpus/with_chords").iterdir():
    score = m21.converter.parse(score_path)
    for element in score.recurse().getElementsByClass(m21.harmony.ChordSymbol):
        element.activeSite.remove(element)
    output_path = output_dir / f"{score_path.name}"
    score.write("musicxml", fp=output_path)
```
